Load summary and inspect

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# Load random attack full-scale summary
# ============================================================

summary_path = Path("random_attack_fullscale_summary.csv")

df = pd.read_csv(summary_path)

print("===== SUMMARY LOADED =====")
print("Shape:", df.shape)
print(df.head())

print("\n===== RATIOS =====")
print(df["malicious_ratio"].value_counts().sort_index())

print("\n===== METHODS =====")
print(df["method"].value_counts())

print("\n===== ENVIRONMENTS =====")
print(df[["envId", "envName"]].drop_duplicates().sort_values("envId"))

print("\n===== FORMATIONS =====")
print(df[["formationType", "formationName"]].drop_duplicates().sort_values("formationType"))
```
